```python
import numpy as np 


def get_mag(x):
    return np.linalg.norm(x)

def get_a(r, v, mu):
    v = get_mag(v) 
    r = get_mag(r) 
    a = ((2/r) -(v**2)/mu)**-1
    return a 

def get_e_vec(v, r, mu): 
    v_mag = get_mag(v)
    r_mag = get_mag(r)
    e  = ((v_mag**2)/mu - 1/r_mag ) * r
    e -= (1/mu) * (np.dot(r,v)) * v
    return e 

def get_angular_momentum_vec(r,v): 
    return np.cross(r,v)

def get_inclination(h, K): 
    cos_i = np.dot(h/np.linalg.norm(h),K) 
    return np.arccos(cos_i)

def get_n(k, h): 
    n = np.cross(k, h)
    return n 

def get_raan(n, h, k, I, J):
    cos_omega = np.dot(n, I)/get_mag(n)
    omega = np.arccos(cos_omega)
    if (np.dot(n, J) < 0): 
        return 2*np.pi - omega
    else: 
        return omega
 

def get_arg_pariapsis(e, n, K): 
    e_mag = get_mag(e)
    n_mag = get_mag(n)
    cos_w = np.dot(n,e)
    cos_w /= (e_mag * n_mag)
    w = np.arccos(cos_w)
    if (np.dot(e, K) < 0): 
        return 2*np.pi - w
    else: 
        return w

def get_true_anomoly(r, e, v): 
    r_mag = get_mag(r)
    e_mag = get_mag(e)
    cos_f = np.dot(r,e)/(r_mag*e_mag)
    f = np.arccos(cos_f)
    if (np.dot(r, v) < 0): 
        return 2*np.pi - f
    else: 
        return f

def get_radial_velocity(r,v): 
    r_direction = r/np.linalg.norm(r)
    v_r = np.dot(r_direction, v) 
    return v_r

```

```python
# Case 1 
K =  np.array([0,0,1])
I = np.array([1,0,0])
J = np.array([0,1,0])

r_0 = np.array([2995.595477195, -4858.411420716, 3338.578470848])
v_0 = np.array([8.232783661, 5.600188238, 2.249982802])
mu_E = 398600

r_mag = get_mag(r_0)
v_mag = get_mag(v_0)
a_1 = get_a(r_0, v_0, mu_E)
e_vec = get_e_vec(v_0, r_0, mu_E)
e_mag = get_mag(e_vec)
h_vec = get_angular_momentum_vec(r_0, v_0)
n = get_n(K, h_vec)
i = get_inclination(h_vec, K=K)
RAAN  = get_raan(n, h_vec, K, I, J)
w = get_arg_pariapsis(e_vec, n, K)
f = get_true_anomoly(r_0, e_vec, v_0)
h_mag = get_mag(h_vec)
v_r = get_radial_velocity(r_0, v_0) 

ans = [r_mag, v_mag, a_1, e_vec, e_mag, h_vec, i, n, RAAN, w, f, h_mag, v_r] 
ans_string = ["r_mag", "v_mag", "a_1", "e_vec", "e_mag", "h_vec", "i", "n", "RAAN", "w", "f", "h_mag", "v_r"]

for i  in range(len(ans)):
    if type(ans[i]) ==  np.float64:  
        print(f"{(ans_string[i])} = {ans[i]:.3f}")
        print(f"\n")
    else: 
        var = ans[i]
        print(f"{(ans_string[i])} = [{var[0]:.3f}, {var[1]:.3f}, {var[2]:.3f},]")
        print(f"\n")
```
