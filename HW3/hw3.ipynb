```python
import numpy as np 


def get_mag(x):
    return np.linalg.norm(x)

def get_a(r, v, mu):
    v = get_mag(v) 
    r = get_mag(r) 
    a = ((2/r) -(v**2)/mu)**-1
    return a 

def get_e_vec(v, r, mu): 
    v_mag = get_mag(v)
    r_mag = get_mag(r)
    e  = ((v_mag**2)/mu - 1/r_mag ) * r
    e -= (1/mu) * (np.dot(r,v)) * v
    return e 

def get_angular_momentum_vec(r,v): 
    return np.cross(r,v)

def get_inclination(h, K): 
    cos_i = np.dot(h/np.linalg.norm(h),K) 
    return np.arccos(cos_i)

def get_n(k, h): 
    n = np.cross(k, h)
    return n 

def get_raan(n, h, k, I, J):
    cos_omega = np.dot(n, I)/get_mag(n)
    omega = np.arccos(cos_omega)
    if (np.dot(n, J) < 0): 
        return 2*np.pi - omega
    else: 
        return omega
 

def get_arg_pariapsis(e, n, K): 
    e_mag = get_mag(e)
    n_mag = get_mag(n)
    cos_w = np.dot(n,e)
    cos_w /= (e_mag * n_mag)
    w = np.arccos(cos_w)
    if (np.dot(e, K) < 0): 
        return 2*np.pi - w
    else: 
        return w

def get_true_anomoly(r, e, v): 
    r_mag = get_mag(r)
    e_mag = get_mag(e)
    cos_f = np.dot(r,e)/(r_mag*e_mag)
    f = np.arccos(cos_f)
    if (np.dot(r, v) < 0): 
        return 2*np.pi - f
    else: 
        return f

def get_radial_velocity(r,v): 
    r_direction = r/np.linalg.norm(r)
    v_r = np.dot(r_direction, v) 
    return v_r

```

```python
# WE first solve for E 
import numpy as np 

def solve_E(M, e, E_initial_guess): 
    found = False 
    E = E_initial_guess
    while not found: 
        derivative = 1 - e * np.cos(E) 
        change =  ((E-e*np.sin(E) - M)/(derivative))
        E = E - change 
        if np.abs(change) <= 1e-15:
            E = E - change
            found = True 
    return E 

def get_h(a,e,mu): 
    return np.sqrt(mu* a * (1-e**2))

def get_f(e, E): 
    f_2 = np.arctan2(np.tan(E/2) * np.sqrt(1+e), np.sqrt(1-e))
    return f_2*2 

def get_r_mag(h, mu, f,e): 
    return ((h**2)/mu) * (1 + e*np.cos(f))**-1

def get_rad_vec(r_mag, omega,theta, i):
    R = np.array([np.cos(theta) * np.cos(omega) - np.cos(i) * np.sin(omega) * np.sin(theta), np.cos(theta) * np.sin(omega) + np.cos(i) * np.sin(omega) * np.sin(theta),  np.sin(i) * np.sin(theta)])
    return R * r_mag

def get_v (h, omega,theta,w, i): 
    return np.array([
        (-mu/h) * (np.cos(omega) * (np.sin(theta)  + e*np.sin(w))  + np.sin(omega) * (np.cos(theta) + e*np.cos(w)) * np.cos(i)), 
        -(mu/h) * (np.sin(omega) * (np.sin(theta) + e*np.sin(w)) - np.cos(omega)*(np.cos(theta) + e*np.cos(w)) * np.cos(i)), 
        (mu/h) * ((np.cos(theta)+ e*np.cos(w)) * np.sin(i))
    ])

```

```python
# Case 1 

a = 6828
e = 0.00732
i = 28.5
omega = 55
w = 20
M = 15 
mu = 398600 
I, J, K = np.array([1,0,0]),  np.array([0,1,0]),  np.array([0,0,1])
E = solve_E(np.radians(M), e, 0.02)
h = get_h(a, e, mu)
f = get_f(e, E)
theta = w + f 
r_mag = get_r_mag(h, mu, f, e)
r_vec = get_rad_vec(r_mag, omega, theta, i)
v_vec = get_v(h, omega, theta, w, i)
v_mag = get_mag(v_vec)

ans = [E, h , f , theta, r_mag, r_vec, v_vec, v_mag] 
ans_string = ['E', 'h' , 'f' , 'theta', 'r_mag', 'r_vec', 'v_vec', 'v_mag']
for i  in range(len(ans)):
    #print(f"{(ans_string[i])} = {ans[i]}")

    if type(ans[i]) ==  np.float64:  
        print(f"{(ans_string[i])} = {ans[i]:.3f}")
        print(f"\n")
    else: 
        var = ans[i]
        print(f"{(ans_string[i])} = [{var[0]:.3f}, {var[1]:.3f}, {var[2]:.3f},]")
        print(f"\n")
        
```

```python
def get_r_dot(v_vec, r_vec): 
    U_r = r_vec/np.linalg.norm(r_vec)
    #U_v = v_vec/np.linalg.norm(v_vec)
    return np.dot(U_r, v_vec) * U_r

def get_E_0(r_0, r_dot, a, mu): 
    tan_E_0_y = np.dot(r_0, r_dot)
    tan_E_0_x = (np.sqrt(a * mu) * (1 - (get_mag(r_0)/a)))**-1 
    return np.arctan2(tan_E_0_y, tan_E_0_x)

def get_a(r, v, mu):
    v = get_mag(v) 
    r = get_mag(r) 
    a = ((2/r) -(v**2)/mu)**-1
    return a 

def solve_E_t(t,t0, n, E_0, e): 
    found = False 
    E = E_0
    var = n* (t-t0) + E_0 - e*np.sin(E_0)
    while not found: 
        derivative = 1 - e * np.cos(E) 
        change =  ((E-e*np.sin(E) - var)/(derivative))
        E = E - change 
        if np.abs(change) <= 1e-15:
            E = E - change
            found = True 
    return E 

def get_e_mag(a, h, mu): 
    return np.sqrt(1 - h**2 / (a* mu))

def get_mean_motion(a, mu): 
    return np.sqrt(mu/a**3)

def get_mean_anomaly(E, e):
    return (E - e * np.sin(E))
```

```python
import matplotlib.pyplot as plt 

r_0 = np.array([-158.9554910, -6087.6241378, 3383.8886543])
v_0 = np.array([7.1033414, 1.3310676, 3.2946236])
mu = 398600
R = 6378 

t0 = 0 
t = 2 * 60 * 60 

a = get_a(r_0, v_0, mu)
h = get_mag(np.cross(r_0, v_0))
r_dot = get_r_dot(v_0, r_0)
E_0 = get_E_0(r_0, r_dot, a, mu)
e = get_e_mag(a, h, mu)
n = get_mean_motion(a, mu)
M = get_mean_anomaly(E_0, e)
time = np.linspace(t0, t, 100)

vects_r, vects_v = [], []
for t in time: 
    E = solve_E_t(t, t0, n, E_0, e)
    F = 1 - (a/get_mag(r_0)) * (1-np.cos(E - E_0))
    G = (t-t0) - np.sqrt(a**3/mu) * ((E-E_0) - np.sin(E- E_0))
    r = F * r_0 + G * v_0 
    r_mag = get_mag(r)
    r_0_mag = get_mag(r_0)
    F_dot = - ((np.sqrt(mu * a))/(r_mag * r_0_mag)) * np.sin(E-E_0)
    G_dot = 1 - (a/r_mag) * (1 - np.cos(E - E_0))
    v = F_dot * r_0 + G_dot*v_0 
    vects_r.append(r)
    vects_v.append(v)

vects_v = np.array(vects_v)
vects_r = np.array(vects_r)
x, y, z = vects_r.T[0], vects_r.T[1], vects_r.T[2]
fig = plt.figure(dpi=300) 
ax = fig.add_subplot(111, projection ="3d")
ax.plot(x, y, z)
ax.set_title("Orbit (tol = 1e-13)")
ax.set_xlabel("X (km)")
ax.set_ylabel("Y (km)")
ax.set_zlabel("Z (km)")



```
